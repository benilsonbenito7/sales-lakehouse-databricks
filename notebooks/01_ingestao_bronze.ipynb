#Camada Bronze

##Objectivo

Ingestao dos dados brutos do arquivo csv e armazena-los em uma tabela delta, preservando assim o conteudo original dos dados para o processamento posterior.

```python
df_bronze = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load("/Volumes/retail_sales/bronze/volume_raw/sales_data_sample-selected-columns(1).csv")
)

#df_bronze.printSchema()
#isplay(df_bronze.limit(10))
#display(df_bronze.count())
```

##Leitura do arquivo CSV completo

Nesta etapa foi realizado a leitura do arquivo csv dos dados das vendas

###Observacoes

1 - O arquivo possui **2.823 resgistros**

2 - Algumas colunas possuem tipos de dados incorectos que seram tratados na camada silver

```python
df_bronze.write.format("delta").mode("append").option("overwriteSchema", True).option("mergeSchema", "true").saveAsTable("retail_sales.bronze.sales_raw")
```

## Salvando os dados em uma Delta Table

Nesta etapa, os dados brutos são armazenados em uma Delta Table.

Os dados são armazenados em uma Delta Table, que utiliza arquivos **Parquet** como formato de armazenamento e adiciona recursos do **Delta Lake**, como controle de versões, transações ACID e gerenciamento do schema.
